```python
# Vectorización con TF-IDF
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(preguntas)
```

```python
# Función para encontrar la mejor coincidencia
def responder_pregunta(consulta_usuario):
  # vectorizo la entrada
  consulta_vec = vectorizer.transform([consulta_usuario])

  similitudes = cosine_similarity(consulta_vec, tfidf_matrix).flatten()
  i = similitudes.argmax() # indice de mejor coincidencia
  print(f'similitud:', i)

  intencion = intenciones[i]
  print(f'intencion:', intencion)

  score = similitudes.max()
  if score < 0.3:
    return "No entendí tu consulta, ¿podés reformularla?", None, score
  return respuestas[intencion], intencion, score
```

```python
# Ejemplo de consulta
consulta = "que precio tiene funcional"
respuesta = responder_pregunta(consulta)
print(respuesta)
```
