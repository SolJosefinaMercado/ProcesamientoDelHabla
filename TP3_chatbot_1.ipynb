```python
# Función para encontrar la mejor coincidencia chat bot TF-IDF
def responder_pregunta(consulta_usuario):
  # vectorizo la entrada
  consulta_vec = vectorizer.transform([consulta_usuario])

  similitudes = cosine_similarity(consulta_vec, tfidf_matrix)
  i = similitudes.argmax() # indice de mejor coincidencia
  print(f'similitud:', i)

  intencion = intenciones[i]
  print(f'intencion:', intencion)

  score = similitudes.max()
  if score < 0.3:
    return "No entendí tu consulta, ¿podés reformularla?", None, score
  return score, respuestas[intencion]
```

Voy a utilizar Spacy porque esta entrenado en el idioma españolm, ademas con otra opcion como gensim deberia bajar aparte un modelo como fastText.

SpaCy tiene vectores de palabras pre entrenados, por lo que espero que entienda la similitud entre palabras como "cuota", "abono" y "mensualidad"

```python
import spacy
import numpy as np

nlp = spacy.load("es_core_news_md")
```

```python
import unicodedata
# separa cada letra de su tilde (NFD) y después descarta las tildes (categoría “Mn”). “Cómo” queda como “como”.
def normalizar(texto):
  texto = texto.lower()
  texto  = unicodedata.normalize('NFKD', texto).encode('ASCII', 'ignore').decode('utf-8')
  return "".join(c for c in texto if unicodedata.category(c) != "Mn")

def vectorizar(texto):
  doc = nlp(normalizar(texto))
  tokens = [t for t in doc if not t.is_stop and not t.is_punct and t.has_vector]

  if not tokens:
    return doc.vector
  return np.mean([t.vector for t in tokens], axis = 0)
```

```python
# en TF-IDF normalizar ayuda siempre, pero en embeddings depende del vocabulario del modelo.
# en este caso las palabras tienen un vector, no se pierde información

for w in ["informacion", "información", "inscripcion", "inscripción", "cómo", "como"]:
    print(w, nlp(w)[0].has_vector)
```
